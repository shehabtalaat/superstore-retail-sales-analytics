## 4. Newly Engineered Features

During the data transformation phase, we derived the following row-level, financial, temporal, and customer-segmentation features:

###  Temporal & Operational Features
* **`Shipping_Days`:** `(Ship Date - Order Date)` — Number of days taken to fulfill and ship the order.
* **`Order_Year`:** `Order Date.dt.year` — The calendar year of the transaction.
* **`Order_Quarter`:** `Order Date.dt.quarter` — The financial quarter (Q1–Q4) of the order.
* **`Order_Month`:** `Order Date.dt.month` — Numerical month of the order (1–12).
* **`Order_Month_Name`:** `Order Date.dt.strftime('%b')` — Abbreviated month name (Jan, Feb, etc.).
* **`Order_Day_Name`:** `Order Date.dt.strftime('%a')` — Day of the week (Mon, Tue, etc.).
* **`Is_Weekend`:** `Order Date.dt.day_of_week >= 5` — Flag indicating whether the purchase occurred on a weekend.

---

###  Financial & Profitability Metrics
* **`Profit_Margin`:** `Profit / Sales` — Transaction-level profit margin ratio.
* **`Original_Sales`:** `Sales / (1 - Discount)` — Estimated gross revenue before discounts.
* **`Unit_Price`:** `Original_Sales / Quantity` — Base unit price of the purchased item.
* **`Discount_Amount`:** `Original_Sales - Sales` — Absolute monetary value given as discount.
* **`Is_Loss`:** `1 if Profit < 0 else 0` — Binary indicator for loss-making transactions.
* **`Discount_Bracket`:** `pd.cut(Discount)` — Binned discount tiers (`No Discount (0%)`, `Low (1-20%)`, `Medium (21-50%)`, `High (>50%)`).

```python
# fn of importing libraries
def import_libraries():
    global pd, np, plt, sns, pa
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    import pyarrow as pa
    print(" All libraries imported successfully!\n")
```

```python
# fn of reading data
def read_data(file_path):
    df= pd.read_csv(file_path)
    print(f"Data shape: {df.shape}\n")
    display(df.head())
    
    return df
```

```python
# fn of inspecting data
def inspect_data(df):
        
    df_raw = df.copy()
    cols_to_drop = ['Unnamed: 0', 'Row ID']
    df_raw = df_raw.drop(columns=cols_to_drop, errors='ignore')
    
    print(" 1. Data Info & Missing Values Summary")
    print("="*50)
    summary_df = pd.DataFrame({
        'Data Type': df_raw.dtypes,
        'Non-Null Count': df_raw.notnull().sum(),
        'Missing Values': df_raw.isnull().sum(),
        'Missing %': (df_raw.isnull().sum() / len(df_raw) * 100).round(2),
        'Unique Values': df_raw.nunique()
    })
    display(summary_df)
    print("\n")
    
    print(" 2. NUMERICAL COLUMNS SUMMARY")
    print("="*50)
    display(df_raw.describe(include=[np.number]))
    print("\n")
    
    print(" 3. CATEGORICAL / STRING COLUMNS SUMMARY")
    print("="*50)
    display(df_raw.describe(include=['object', 'category']))
    print("\n")

    print(" 4. DATASET OVERVIEW & DUPLICATES")
    print("="*50)
    print(f"• Total Rows: {df_raw.shape[0]}")
    print(f"• Total Columns: {df_raw.shape[1]}")
    print(f"• Duplicate Rows: {df_raw.duplicated().sum()}")
    print("\n")
    
    return df_raw
```

```python
# fn of checking categorical columns
def check_categories(df):
    
    selected_cols = ['Discount', 'Category', 'Region', 'Segment', 'Ship Mode']
    
  
    print(" 5. The most important categorical columns and their unique values:")
    print("="*50)
    
    for col in selected_cols:
        if col in df.columns:
            print(f"\n Column: [{col}] — ({df[col].nunique()} Unique Values)")
            print("-" * 40)
            
            counts = df[col].value_counts(dropna=False)
            percentages = (df[col].value_counts(dropna=False, normalize=True) * 100).round(2)
            
            val_df = pd.DataFrame({
                'Count': counts,
                'Percentage (%)': percentages
            })
            display(val_df)
        else:
            print(f"the column [{col}] does not exist in the DataFrame.\n")

          
```

```python
# fn of checking nulls and duplicates
def check_nulls_and_duplicates(df):
    # 1. Check missing values column by column
    
    print("1. MISSING VALUES (NULLS) BY COLUMN")
    print("=" * 50)
    
    found_nulls = False
    for col in df.columns:
        null_rows = df[df[col].isnull()]
        if len(null_rows) > 0:
            found_nulls = True
            print(f"\n Column [{col}] has {len(null_rows)} missing row(s):")
            display(null_rows)
            
    if not found_nulls:
        print(" No missing values (nulls) found in any column!")

    # 2. Check duplicate rows
    print(" 2. DUPLICATE ROWS")
    print("=" * 50)
    
    duplicates = df[df.duplicated(keep=False)]
    if len(duplicates) > 0:
        print(f" Found {len(duplicates)} duplicate row(s):")
        display(duplicates)
    else:
        print(" No duplicate rows found!")
```

```python
# fn of changing column type
def change_column_type(df, col_name, new_type):
    
    if col_name in df.columns:
        df[col_name] = df[col_name].astype(new_type)
        print(f" Column [{col_name}] type successfully changed to -> {new_type}")
    else:
        print(f" Column [{col_name}] not found in DataFrame!")
    return df
```

```python
# fn of changing date column type
def change_date_type(df, col_name, date_format=None):
 
    if col_name in df.columns:
        df[col_name] = pd.to_datetime(df[col_name], format=date_format, errors='coerce')
        print(f" Column [{col_name}] successfully converted to datetime64[ns]")
    else:
        print(f" Column [{col_name}] not found in DataFrame!")
    return df
```

```python
# fn of handling missing values
def handle_missing_value(df, col_name, fill_value):

    if col_name in df.columns:
        null_count_before = df[col_name].isnull().sum()
        
        if null_count_before > 0:
            df[col_name] = df[col_name].fillna(fill_value)
            print(f" Successfully filled {null_count_before} missing value(s) in [{col_name}] with -> '{fill_value}'")
        else:
            print(f"ℹ Column [{col_name}] has no missing values to fill.")
    else:
        print(f" Column [{col_name}] not found in DataFrame!")
        
    return df
```

```python
# fn of removing duplicates
def remove_duplicates(df):
    duplicates_count = df.duplicated().sum()
    print(f"Original shape: {df.shape}")
    if duplicates_count > 0:
        df = df.drop_duplicates()
        print(f" Successfully removed {duplicates_count} duplicate row(s)!")
    else:
        print(" No duplicate rows found to remove.")
    print(f"Shape after removing duplicates: {df.shape}")
    return df
```

```python
# fn of checking and fixing consistency in categorical columns
def check_and_fix_consistency(df, cat_cols):
    print(" 1. CLEANING WHITESPACES & NORMALIZING STRINGS")
    print("=" * 50)
    
    for col in cat_cols:
        if col in df.columns:
            if df[col].dtype.name == 'category':
                df[col] = df[col].cat.rename_categories(lambda x: str(x).strip())
            else:
                df[col] = df[col].astype(str).str.strip()
                
    print(" Whitespaces cleaned successfully across categorical columns!\n")
    
    print(" 2. CHECKING UNIQUE VALUES FOR EACH CATEGORICAL COLUMN")
    print("=" * 50)
    
    for col in cat_cols:
        if col in df.columns:
            unique_vals = df[col].unique()
            print(f"\n Column: [{col}] — ({len(unique_vals)} Unique Values)")
            print("-" * 40)
            print(unique_vals)
            
    return df
```

```python
# fn of detecting outliers using IQR method
def detect_outliers_iqr(df, num_cols):

    print(" 1. OUTLIERS COUNT USING IQR METHOD")
    print("=" * 50)
    
    for col in num_cols:
        if col in df.columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            
            outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
            
            print(f" Column: [{col}]")
            print(f"   • Range Bounds: Lower = {lower_bound:.2f} | Upper = {upper_bound:.2f}")
            print(f"   • Outliers Count: {len(outliers)} ({len(outliers)/len(df)*100:.2f}%)\n")

    plt.figure(figsize=(15, 6))
    for i, col in enumerate(num_cols, 1):
        plt.subplot(1, len(num_cols), i)
        sns.boxplot(y=df[col], color='skyblue')
        plt.title(f'Boxplot of {col}', fontsize=12)
        plt.ylabel('')
        
    plt.tight_layout()
    plt.show()
```

```python
# fn of adding derived row-level features
def add_features(df):

    print("  CREATING DERIVED ROW-LEVEL FEATURES")
    print("=" * 50)
    
    df['Order_Year'] = df['Order Date'].dt.year
    df['Order_Quarter'] = df['Order Date'].dt.quarter
    df['Order_Month'] = df['Order Date'].dt.month
    df['Order_Month_Name'] = df['Order Date'].dt.strftime('%b')
    df['Order_Day_Name'] = df['Order Date'].dt.strftime('%a')
    df['Is_Weekend'] = df['Order Date'].dt.day_of_week >= 5
    
    df['Shipping_Days'] = (df['Ship Date'] - df['Order Date']).dt.days
    df['Profit_Margin'] = df['Profit'] / df['Sales']

    df['Original_Sales'] = df['Sales'] / (1 - df['Discount'])
    df['Unit_Price'] = df['Original_Sales'] / df['Quantity']

    df['Discount_Amount'] = df['Original_Sales'] - df['Sales']

    df['Is_Loss'] = np.where(df['Profit'] < 0, 1, 0)
    bins = [-0.01, 0.0, 0.2, 0.5, 1.0]
    labels = ['No Discount (0%)', 'Low (1-20%)', 'Medium (21-50%)', 'High (>50%)']
    df['Discount_Bracket'] = pd.cut(df['Discount'], bins=bins, labels=labels)
 
    return df
```

```python
# fn of validating new features
def validate_new_features(df):

    print(" STEP 4: FEATURE VALIDATION CHECK")
    print("=" * 50)
    
    new_cols = [
        'Order_Year', 'Order_Quarter', 'Shipping_Days', 
        'Original_Sales', 'Discount_Amount', 'Unit_Price', 
        'Profit_Margin', 'Is_Loss', 'Discount_Bracket', 'Is_Weekend'
    ]
    #null
    null_counts = df[new_cols].isnull().sum()
    print(" Null Values Count in New Features:")
    print(null_counts[null_counts > 0] if (null_counts > 0).any() else " No Null/NaN values found!")
    print("-" * 40)
    
    # inft
    num_cols = df[new_cols].select_dtypes(include=['float64', 'int64']).columns
    inf_counts = np.isinf(df[num_cols]).sum()
    print(" Infinite Values Count (e.g. Division by zero):")
    print(inf_counts[inf_counts > 0] if (inf_counts > 0).any() else " No Infinite values found!")
    print("-" * 40)
    
    # negative shipping days
    negative_shipping = (df['Shipping_Days'] < 0).sum()
    print(f" Negative Shipping Days Count: {negative_shipping} " + (" (OK)" if negative_shipping == 0 else " (Fix Needed!)"))
    
    
    print("\n All features validated and safe for Exploratory Data Analysis (EDA)!")


```

## Reusable Functions & Pipeline Documentation

---

### 1. `import_libraries()`
* **Description:** Imports all essential Python libraries required for data manipulation, numerical operations, visualizations, and file storage (`pandas`, `numpy`, `matplotlib.pyplot`, `seaborn`, `pyarrow`)

### 2. `read_data(file_path)`
* **Description:** Loads the raw dataset from a given file path, prints its dimensional structure (`Data shape`), and displays the first 5 records (`head`).

### 3. `inspect_data(df)`
* **Description:** Conducts a comprehensive exploratory data health check, including data types, missing value percentages, descriptive statistics for numerical and categorical features, and duplicate row detection (excluding non-informative columns such as `Row ID` and `Unnamed: 0`).

### 4. `check_categories(df)` *(or `check_selected_categories(df)`)*
* **Description:** Summarizes unique value counts, frequencies, and percentage distributions for key categorical variables (`Category`, `Segment`, `Region`, `Ship Mode`, `Discount`).

### 5. `check_nulls_and_duplicates(df)`
* **Description:** Performs a targeted inspection to extract and display all rows containing missing values (`Nulls`) per column, as well as detecting and outputting fully duplicated records.

### 6. `change_column_type(df, col_name, new_type)`
* **Description:** Safely casts a specified column to a new target data type (such as converting string columns to `category` or `string`) after verifying column existence.

### 7. `change_date_type(df, col_name, date_format=None)`
* **Description:** Converts text-based date columns to `datetime64[ns]` format using `pd.to_datetime` to enable time-series operations and temporal feature extraction.

### 8. `handle_missing_value(df, col_name, fill_value)`
* **Description:** Imputes missing values (`Nulls`) in a specified column with a designated static value and logs the number of filled instances.

### 9. `remove_duplicates(df)`
* **Description:** Drops duplicate rows across the DataFrame to ensure data integrity and outputs dataset dimensions before and after cleaning.

### 10. `check_and_fix_consistency(df, cat_cols)`
* **Description:** Standardizes categorical text fields by stripping leading/trailing whitespaces, normalizing string formats, and inspecting unique values across target categories.

### 11. `detect_outliers_iqr(df, num_cols)`
* **Description:** Identifies and quantifies outliers in numerical features using the Interquartile Range (`IQR`) method, calculating lower/upper bounds and generating corresponding `Boxplots`.

### 12. `add_features(df)`
* **Description:** Derives essential time-based attributes (`Order_Year`, `Order_Quarter`, `Order_Month`, `Shipping_Days`, `Is_Weekend`)[cite: 1] and financial performance metrics (`Profit_Margin`, `Original_Sales`, `Unit_Price`, `Discount_Amount`, `Is_Loss`, `Discount_Bracket`).

### 13. `validate_new_features(df)`
* **Description:** Validates engineered features to ensure data quality, verifying the absence of `Null` values, infinite values (`Inf`), and negative shipping duration anomalies.

```python
# Relation between Category and Sub-Category with Sales, Profit, Profit Margin, and Loss Rate
category_perf = df_clean.groupby(['Category', 'Sub-Category']).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Margin=('Profit_Margin', 'mean'),
    Loss_Rate=('Is_Loss', 'mean')
).reset_index().sort_values(by='Total_Profit', ascending=False)
category_perf
```

```python
# Grouping by Discount Bracket to analyze performance metrics
discount_perf = df_clean.groupby('Discount_Bracket', observed=False).agg(
    Order_Count=('Order ID', 'count'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Margin=('Profit_Margin', 'mean'),
    Loss_Rate=('Is_Loss', 'mean')
).reset_index()
discount_perf
```

```python
# Grouping by Ship Mode to analyze performance metrics
shipmode_perf = df_clean.groupby('Ship Mode').agg(
    Total_Orders=('Order ID', 'count'),
    Avg_Shipping_Days=('Shipping_Days', 'mean'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Margin=('Profit_Margin', 'mean')
).reset_index().sort_values(by='Total_Orders', ascending=False)
shipmode_perf
```

```python
# Grouping by Year and Quarter to analyze performance metrics
quarterly_perf = df_clean.groupby(['Order_Year', 'Order_Quarter']).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).reset_index()
quarterly_perf
```

```python
# Grouping by Order ID and Segment to analyze performance metrics
order_level = df_clean.groupby(['Order ID', 'Segment']).agg(
    Order_Sales=('Sales', 'sum'),
    Order_Profit=('Profit', 'sum')
).reset_index()
order_level
```

```python
# Grouping by Segment to analyze performance metrics
segment_perf = order_level.groupby('Segment').agg(
    Total_Orders=('Order ID', 'count'),
    AOV=('Order_Sales', 'mean'),
    Avg_Profit_Per_Order=('Order_Profit', 'mean'),
    Total_Sales=('Order_Sales', 'sum'),
    Total_Profit=('Order_Profit', 'sum')
).reset_index().sort_values(by='Total_Profit', ascending=False)
segment_perf 
```

## 5. Customer Segmentation (RFM Analysis)

```python
# RFM Analysis
recency_reference = df_clean["Order Date"].max() + pd.Timedelta(days=1)
recency_reference

RFM=df_clean.groupby("Customer ID").agg(
    recency = ("Order Date", lambda x: (recency_reference - x.max()).days),
    frequency = ("Order ID", "nunique"),
    monetary_value = ("Sales", "sum")
).reset_index()
RFM
```

```python
# RFM Scoring
RFM['R_Score'] = pd.qcut(RFM['recency'], q=4, labels=[4, 3, 2, 1])
RFM['F_Score'] = pd.qcut(RFM['frequency'].rank(method='first'), q=4, labels=[1, 2, 3, 4])
RFM['M_Score'] = pd.qcut(RFM['monetary_value'], q=4, labels=[1, 2, 3, 4])

RFM['RFM_Total'] = RFM['R_Score'].astype(int) + RFM['F_Score'].astype(int) + RFM['M_Score'].astype(int)

def assign_segment(score):
    if score >= 10:
        return 'A'
    elif score >= 8:
        return 'B'
    elif score >= 6:
        return 'C'
    else:
        return 'D'
```

```python
RFM['Customer_Segment'] = RFM['RFM_Total'].apply(assign_segment)
RFM.head()
```

```python
# RFM Summary by Customer Segment
RFM.groupby('Customer_Segment').agg({
    'recency': 'mean',
    'frequency': 'mean',
    'monetary_value': 'mean',
    'Customer ID': 'count'
}).rename(columns={'Customer ID': 'Customer_Count'}).round(2).reset_index()
```
